```python
import fitparse
import pandas as pd
import matplotlib.pyplot as plt

# 1. Point directly to the file in your working directory
fit_file_path = '24497115516_ACTIVITY.fit'

# 2. Parse the binary FIT file
fitfile = fitparse.FitFile(fit_file_path)

records = []
for record in fitfile.get_messages('record'):
    data = {}
    for record_data in record:
        data[record_data.name] = record_data.value
    records.append(data)

# 3. Convert to a Pandas DataFrame
df = pd.DataFrame(records)
print("Columns available in this file:", df.columns.tolist())

# 4. Save as CSV
csv_output_path = 'converted_activity_data.csv'
df.to_csv(csv_output_path, index=False)
print(f"Successfully converted and saved to {csv_output_path}!")

# 5. Quick Visualization of Heart Rate over time
if 'heart_rate' in df.columns and 'timestamp' in df.columns:
    plt.figure(figsize=(10, 4))
    plt.plot(df['timestamp'], df['heart_rate'], color='purple', linewidth=1.2)
    plt.title('Heart Rate Activity Sample')
    plt.xlabel('Timestamp')
    plt.ylabel('Heart Rate (bpm)')
    plt.grid(True)
    plt.tight_layout()
    plt.show()
else:
    print("Heart rate column not found, displaying the first few rows:")
    display(df.head())
```

```python
import fitparse
import pandas as pd
import matplotlib.pyplot as plt

# 1. Point directly to the file in your working directory
fit_file_path = '24519659519_ACTIVITY.fit'

# 2. Parse the binary FIT file
fitfile = fitparse.FitFile(fit_file_path)

records = []
for record in fitfile.get_messages('record'):
    data = {}
    for record_data in record:
        data[record_data.name] = record_data.value
    records.append(data)

# 3. Convert to a Pandas DataFrame
df = pd.DataFrame(records)
print("Columns available in this file:", df.columns.tolist())

# 4. Save as CSV
csv_output_path = 'converted_activity_data.csv'
df.to_csv(csv_output_path, index=False)
print(f"Successfully converted and saved to {csv_output_path}!")

# 5. Dynamic Visualization of alternative metrics (Cadence, Speed, or Altitude)
if 'timestamp' in df.columns:
    # Let's check for alternative features in order of preference
    target_col = None
    label_name = ""
    color_choice = "purple"

    if 'cadence' in df.columns:
        target_col = 'cadence'
        label_name = 'Cadence (rpm / steps per min)'
        color_choice = 'teal'
    elif 'enhanced_speed' in df.columns:
        target_col = 'enhanced_speed'
        label_name = 'Speed (m/s)'
        color_choice = 'orange'
    elif 'altitude' in df.columns:
        target_col = 'altitude'
        label_name = 'Altitude (m)'
        color_choice = 'green'
    elif len(df.columns) > 1:
        # Fallback to the second column if none of the above match
        target_col = df.columns[1]
        label_name = target_col.replace('_', ' ').title()
        color_choice = 'royalblue'

    if target_col:
        plt.figure(figsize=(10, 4))
        plt.plot(df['timestamp'], df[target_col], color=color_choice, linewidth=1.2)
        plt.title(f'{label_name} Over Time')
        plt.xlabel('Timestamp')
        plt.ylabel(label_name)
        plt.grid(True)
        plt.tight_layout()
        plt.show()
    else:
        print("Not enough numeric sequence data to plot, displaying the first few rows:")
        display(df.head())
else:
    print("Timestamp column not found, displaying the first few rows:")
    display(df.head())
```
